### **7. Data Types**

Every value in Python has a datatype. Since everything is an object in Python programming, data types are actually classes and variables are instance (object) of these classes.

Below are few of the commonly used data types

### **Python Numbers**

**Integers, floating point numbers and complex numbers** fails under Python numbers category. They are defined as int, float and complex class in Python, and the **isinstance()** function to check if an object belongs to a particular class.

```python
data1 = 12
print(data1, "is type of", type(data1)) # integer data type
```

```python
data2 = 12.3
print(data2, "is type of", type(data2)) # float data type
```

```python
data3 = 10 + 12j
print(data3, "is type of", type(data3)) # data type is changed to complex number
```

### **Boolean**

Boolean represents True or False values

```python
data4 = True
print(data4, "is type of", type(data4)) # boolean data type
```

```python
print('Data3', 'is of type complex?', (data3, complex))
print('Data3', 'is of type complex?', isinstance(data3, complex))
```

### **Python Strings**

Srings is sequence of Unicode characters. We can use single quotes or double quotes to represent strings.
Muti-line strings can be denoted using triple quotes (single/double), **"or".**

A string in Python consists of a series or <u>sequence of chacters</u> - **letters, numbers, and special characters.**

Strings can be indexed - often synonymously called subscripted as well.

The first character of a string has the index 0.

```python
stringVariable = "This is a string"

print('stringVariable is:', stringVariable)
print('First character in stringVariable is:', stringVariable[0])
```

```python
# Slicing of String variable
print(stringVariable[0:7]) # This will print first 7 characters of stringVariable
print(stringVariable[:7]) # This will print characters from index 0 to 6 of stringVariable
print(stringVariable[5:12]) # This will print characters from index 5 to 11 of stringVariable
print(stringVariable[5:16]) # This will print characters from index 5 to end 16 of stringVariable
print(stringVariable[:-1]) # This will print characters from index 0 to second last character of stringVariable
print(stringVariable[-1:]) # This will print characters from index -1 to end of stringVariable
```

### **Python List**

List is an <u>ordered sequence</u> of items. It is one of the most used datatype in Python and is ver flexible.

All the items in a list do not need to be of the same type.

Declaring a list is pretty straight forward. Items separated by commas are closed within square brackets **'[]'**

```python
mylist = [5, 2.2, 'Third Element', 14, 75, "End"]
print(mylist)
```

```python
mylist[3] # This will print 4th element of list which is 14
```

```python
mylist[2] # This will print 3rd element of list which is 'Third Element'
```

Lists are mutable, which means, value of elements of a list can be changed

```python
mylist[2] = "ElementeTheree" # This will change 3rd element of list to 'ElementeTheree'
print(mylist)
```

```python
mylist.remove(14) # This will remove 14 from list
print(mylist)
```

```python
mylist.pop(1) # This will remove element at index 1 which is 2.2
print(mylist)
```

```python
mylist.append(2.2) # This will add 2.2 at the end of list
print(mylist)
```

### **Python Tuple**

Tuple is an <u>ordered sequence</u> of items same as list.

The only difference is that tuples are <u>immutable</u>. **Tuples once created cannot be modified.**

Tuples are used to write-protect data and are usually faster list as it cannot change dynamically.

It is defined within parentheses () where items are separated by commas.

```python
myTuple = (10, 20, 30, "Text")
print(myTuple)
```

```python
print(myTuple[2]) # This will print 3rd element of tuple which is 30
```

```python
myTuple[2] = 25 # This will give error because tuple is immutable and we cannot change its elements
print(myTuple)
```
