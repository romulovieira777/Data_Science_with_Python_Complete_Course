## **6. Variables**

A variable is an alias given to a location in memory that is used to store some data (value).

Each data point stored in the memory is given unique name to differentiate between different memory locations (and thus data points).

**The rules for writing a variable name is same as the rules for writing identifiers in Python.**

We don't need to declare a variable before using it. In Python, we simply assign a value to a variable and it will exist. We don't even have to declare the **type** of the variable. This is handled internally acoording to the type of value we assign to the variable.

#### **Variable assignements**

We use the assignment operator (=) to assign values to a variable.

```python
IntergerVariable = 4
DecimalVariable = 22.2
StringVariable = "StringValue"

print(IntergerVariable)
print(DecimalVariable)
print(StringVariable)
```

#### **Multiple assignements**

Use comma to separate the variables and their corresponding values.

```python
IntergerVariable, DecimalVariable, StringVariable = 4, 22.2, "StringValue"

print(IntergerVariable)
print(DecimalVariable)
print(StringVariable)
```

To initialize multiple variables with same value use 1 assignment.

```python
IntergerVariable = DecimalVariable = StringVariable = 10

print(IntergerVariable)
print(DecimalVariable)
print(StringVariable)
```

#### **Storage Locations**

```python
print(id(IntergerVariable))
```

```python
print(id(DecimalVariable))
```

```python
print(id(StringVariable))
```

**Interger1** and **Interger2** points to same memory location in the below example.
